```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
#generate synthetic data
np.random.seed(42) #to get same set of random data
X=np.random.rand(100,1)*100
Y=3*X+np.random.randn(100,1)*2
```

```python
#split data
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
#fit linear regression
model=LinearRegression()
model.fit(X_train,Y_train) #training data and target
```

```python
#make prediction
y_pred=model.predict(X_test)#based on testing data set
```

```python
#print coefficients
print("slope:",model.coef_[0][0])
print("intercept:",model.intercept_[0])
```

```python
print("slope:",model.coef_.item())
print("intercept:",model.intercept_.item())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X_test,Y_test,color="blue",label="Actual")
plt.plot(X_test,y_pred,color="red",label="prediction")
plt.title("linear regression model")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
#evaluate performance
mse=mean_squared_error(Y_test,y_pred)
r2=r2_score(Y_test,y_pred)
print("mse:",mse)
print("r-sqaured:",r2)
```
